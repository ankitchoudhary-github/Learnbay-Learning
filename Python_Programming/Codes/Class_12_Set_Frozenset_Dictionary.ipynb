```python
# Dictionary Properties
# 1. Dictionary is mutable
st_details = {1:{'st_name':'Aniket',
                 'st_age':15,
                 'st_std':'IX',
                 'st_marks':98.89},
              2:{'st_name':'Simran',
                 'st_age':16,
                 'st_std':'X',
                 'st_marks':98.89}}

print('Before Operation')
print(st_details, id(st_details))
st_details[2]['st_std'] = 'XI'
print('After Operation')
print(st_details, id(st_details))
```

```python
# 2. Dictionary is not a sequence - indexing, slicing, concatenation and repetition is not allowed
# 1. membership operation - will be valid for keys only, not for values
# i.e you may check whether key is an element in the dictioanry or not but for values you can't

st_details = {1:{'st_name':'Aniket',
                 'st_age':15,
                 'st_std':'IX',
                 'st_marks':98.89},
              2:{'st_name':'Simran',
                 'st_age':16,
                 'st_std':'X',
                 'st_marks':98.89}}

print('st_name' in st_details)
print(1 in st_details)
print('st_name' in st_details[2])
print('Aniket' in st_details)
```

```python
# Dictionary is an iterable - When you perform iteration on dictionary object, then it always iterates over keys
for ele in st_details:
    for key in st_details[ele]:
      print(ele, key, st_details[ele][key])
```

```python
# 3. Keys inside a dictionary should always be an immutable object, whereas a value can be anything.

st_details = {1:{'st_name':'Aniket',
                 'st_age':15,
                 'st_std':'IX',
                 'st_marks':98.89},
              [2]:{'st_name':'Simran',
                 'st_age':16,
                 'st_std':'X',
                 'st_marks':98.89}}

st_details = {1:{'st_name':'Aniket',
                 'st_age':15,
                 'st_std':'IX',
                 'st_marks':98.89},
              (2,):{'st_name':'Simran',
                 'st_age':16,
                 'st_std':'X',
                 'st_marks':98.89}}
```

```python
# zip(ietrable1, iterable2, iterable3...) - zip object
# zip(*iterables, strict=False) --> Yield tuples until an input is exhausted.
#  |
#  |     >>> list(zip('abcdefg', range(3), range(4)))
#  |     [('a', 0, 0), ('b', 1, 1), ('c', 2, 2)]
#  |
#  |  The zip object yields n-length tuples, where n is the number of iterables
#  |  passed as positional arguments to zip().  The i-th element in every tuple
#  |  comes from the i-th iterable argument to zip().  This continues until the
#  |  shortest argument is exhausted.
#  |
#  |  If strict is true and one of the arguments is exhausted before the others,
#  |  raise a ValueError.
```
