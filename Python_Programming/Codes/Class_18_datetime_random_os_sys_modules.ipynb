```python
# In datetime Module

# date class - for working with just year, month and days
# time class - for working with just Hour, minute, second and microsecond
# datetime class - both date and time - Most commonly used class
# timedelta class - represent a duration or the difference between the datetime objects.
```

```python
import datetime

# Getting current date and time
now = datetime.datetime.now()
print(f'Current date and time is {now}')

# 2026, 1, 1, 10, 30, 30
# # Creating a specific date and time
# specific_time = datetime.datetime(2026, 1, 1, 10, 30, 30)
# print(f'Specific date and time is {specific_time}', type(specific_time))

# Accessing individual component
print(f"Year : {now.year}")
print(f"Month : {now.month}")
print(f"Day : {now.day}")
print(f"Day of the week (Monday = 0 and Sunday = 6): {now.weekday()}")
```

```python
# Formatting of dates and Times(strftime and strptime)
# strftime - String format time - Converts a datetime pobject to a datetime string object
# strptime - String parse time - Converts a date/time string to a datetime object

# %Y - Full Year - 2025
# %m - Month as number - 01-12
# %d - day of the month - 01 - 31
# %H - Hour - 24-Hour Clock
# %M - Minuts - 00-59
# %S - Second - 00-59
# %A - Full Weekday name - Tuesday
# %B - FULL Month name - November
# %p - AM/PM
```
