```python
# 1. Numerical Literal - Integers
100, 200, -300, 0, -5 # - Integers Literal

# Whole Numbers - Any number without decimal point
# age = 10   # Decimal
# letters_age = 'ten'
# roman_age = X

# Only Integers can be represented in different format in Python - Binary, octal, Hexadecimal, Decimal
# Decimal Representation - allows 10 digits - 0-9
# Binary representation -
  # Bi - Two -
  # Allows only two digits(0 and 1),
  # Binary numbers should always starts with 0b or 0B,
  # bin(num) - num can be in any format and this function converts any numbers to binary format

# Octal representation -
  # octa - Eight -
  # Allows only eight digits(0 to 7),
  # OCtal numbers should always starts with 0o or 0O,
  # oct(num) - num can be in any format and this function converts any numbers to octal format

# Hexadecimal representation -
  # hexa - sixteen -
  # Allows only sinxteen digits(0 to 9 and a/A, b/B, c/C, d/D, e/E, f/F),
  # Hexadecimal numbers should always starts with 0x or 0X,
  # hex(num) - num can be in any format and this function converts any numbers to Hexadecimal format
```

```python
decimal_num = 42
print('Decimal Number: ',decimal_num)
binary_num = bin(decimal_num)
print('Binary Representation: ', binary_num)

octal_num = oct(decimal_num)
print('Octal Representation: ', octal_num)

hexadecimal_num = hex(decimal_num)
print('Hexadecimal Representation: ', hexadecimal_num)
```
