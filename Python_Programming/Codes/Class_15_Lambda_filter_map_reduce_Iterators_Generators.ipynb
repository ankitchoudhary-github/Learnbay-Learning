```python
# map - is used in case if you have any iterable and you want to do same operation on each element of
# the iterable object
# return type is map object - iterator

# map(func_name, iterable)

# Convert each value in INR - multiply each element by 97

prices_usd = [10, 25, 4.5, 99.99]

# def usd_to_inr(usd):
#   return usd*97

# list(map(usd_to_inr, prices_usd))

list(map(lambda usd:usd*97, prices_usd))
```

```python
# filter - is used in case if you have any iterable and you want to filter few elements based on some coindition of the iterable object
# return type is filter object(iterator)

# filter(func, iterable)

numbers = list(range(1, 11))
# numbers

list(filter(lambda num:num%2 == 0, numbers))
```

```python
# sum([10, 20, 30, 40])
# max((1, 2, 100, 29, 90, 99, 999))
# min((1, 2, 100, 29, 90, 99, 999))


# reduce - aggregation

from functools import reduce

# reduce(...)
#     reduce(function, iterable[, initial]) -> value

#     Apply a function of two arguments cumulatively to the items of an iterable, from left to right.

#     This effectively reduces the iterable to a single value.  If initial is present,
#     it is placed before the items of the iterable in the calculation, and serves as
#     a default when the iterable is empty.

#     For example, reduce(lambda x, y: x+y, [1, 2, 3, 4, 5], 0)
#     calculates ((((1 + 2) + 3) + 4) + 5).
```

```python
lst = [(1, 9), (0, 7), (8, 7), (6, 7), (4, 7), (2, 4), (5, 9), (9, 0)]

# Find out the product of sum of each tuple in lst

reduce(lambda num1,num2:num1*num2, map(lambda tpl:sum(tpl), lst), 0)
```

```python
# Closure - An outer function return the reference of inner function then it is called closure

# Closure is used to acces or call the nested function objects from out of scope


def outer_function():
  print('Outer Function')
  def inner_function():
    print('Inner Function')
    print('This is Python class')
  return inner_function
my_func = outer_function()
print('----From Global----')
my_func()
```
