```python
# Counting Positive and Negative Stock Movements

# Problem: You have a list of daily stock price changes (positive for increase, negative for decrease)
# and need to count how many days the stock price went up (positive change) and how many days
# it went down (negative change).

daily_changes = [0.5, -0.2, 1.1, 0.0, -0.7, 0.3, -0.1]
pos_change_count = 0
neg_change_count = 0
for change in daily_changes:
  if change > 0:
    pos_change_count += 1
  elif change < 0:
    neg_change_count +=1
  else:
    continue
print(f'Total Positive count is {pos_change_count}')
print(f'Total Negative count is {neg_change_count}')
```

```python
temp_reading = [27, 28, 27.9, 28.1, 25.9, 80, 26, 27, 28]
# if temp_reading goes beyond 30 then processes should stop immediately
for temp in temp_reading:
  if temp <= 30:
    print(f'temperature is {temp}')
    print('Do step 1')
    print('Do step 2')
    print('Do step 3')
  else:
    print(f'temperature is {temp}')
    print('stopping the process as it may be fatal!!!')
    break
print('Outside of for loop')
```
