```python
# 1. Importing a specific function from a module
from store.products import get_product_details

# 2. Importing the whole module
from store import cart

# 3. Importing from a sub-package (Deep nesting)
from store.checkout.payment import process_credit_card

# main.py
# print("Welcome to the E-commerce App")
# print(cart.add_to_cart('Laptop'))

def run_app():
    # Using the imported function
    print(get_product_details())

    # Using the imported module (requires module_name.function_name)
    print(cart.add_to_cart("Laptop"))

    # Using the sub-package function
    print(process_credit_card(999))

if __name__ == "__main__":
    run_app()
```
