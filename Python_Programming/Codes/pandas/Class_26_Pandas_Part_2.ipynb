### Handling the Missing Value

```python
weather_data = pd.read_csv('datasets/ny_weather.csv')
weather_data
```

```python
weather_data.info()
```

```python
weather_data.isnull()
```

```python
weather_data.isnull().sum()
```

```python
weather_data.dropna()
```

```python
weather_data.fillna(0)
```

```python
weather_data[['Temperature', 'DewPoint', 'Humidity', 'Sea Level PressureIn', 'VisibilityMiles', 'WindSpeedMPH']] = weather_data[['Temperature', 'DewPoint', 'Humidity', 'Sea Level PressureIn', 'VisibilityMiles', 'WindSpeedMPH']].fillna(0)
```

```python
weather_data['Event'] = weather_data['Event'].fillna('No EVent')
weather_data
```

```python
weather_data.fillna({'Temperature' : 0,
                         'DewPoint' : 0,
                         'Humidity' : 0,
             'Sea Level PressureIn' : 0,
                  'VisibilityMiles' : 0,
                     'WindSpeedMPH' : 0,
                            'Event' : "No Event"})
```

```python
weather_data['Temperature'].interpolate()
```

### Changing or Adding the Data

```python
df = pd.read_csv('datasets/test.csv')
df
```

```python
df["Total"] = df["Price"] * df["Quantity"]
print(df)
```

```python
df["GST"] = df["Total"] * 0.18
df["Total_with_GST"] = df["Total"] + df["GST"]
print(df)
```

```python
def price_label (price):
    if price <100:
        return "Cheap"
    else: 
        return "Costly"
```

```python
df ["PriceType"] = df["Price"].apply(price_label)
df
```

```python
# df ["PriceType"] = df["Price"].apply(lambda price: 'Cheap' if price < 100 else 'Costly')
```

```python
df["CategoryShort"] = df["Category"].map({
    "stationery": "STAT",
    "electronics": "ELEC"
})
df
```

```python
df["Category"] = df["Category"].replace("stationery", "office_supplies")

df
```

```python
df["NameUpper"] = df["Name"].str.upper()
df["NameLength"] = df["Name"].str.len()
df["HasO"] = df["Name"].str.contains("o")
print(df)
```

### Dropping the Column

```python
df = df.drop(columns=["CategoryShort", "HasO"])
df
```
