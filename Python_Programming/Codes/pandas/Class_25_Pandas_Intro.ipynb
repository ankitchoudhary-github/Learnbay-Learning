```python
# We want same columns values for ride_3 and ride_5
new_citi_bike_data.loc[['ride_3', 'ride_5'],['start station name', 'starttime', 'bikeid']]
```

```python
new_citi_bike_data.iloc[[2, 4], [1, 2, 0]]
```

```python
# Conditional Indexing
new_citi_bike_data[new_citi_bike_data['tripduration']>200]
```

```python
# In our original datafrmae, we want to access all the trips where start station name is Grand St
citi_bike_data[citi_bike_data['start station name'] == 'Grand St']
```

```python
# Access the trip data for first 8 columns for all the rides where tripduration is more than 200

new_citi_bike_data[new_citi_bike_data['tripduration']>200].iloc[::, :8]
```

```python
new_citi_bike_data.loc[new_citi_bike_data['tripduration']>200, 'tripduration':'bikeid']
```
