# Bivariate Analysis

```python
# When both variables are numerical data


sns.jointplot(x = 'Attack Point',
              y = 'Defense Point',
              data = pokemon)
```

```python
sns.jointplot(x = 'Attack Point',
              y = 'Defense Point',
              data = pokemon,
              kind = 'hex')
```

```python
sns.jointplot(x = 'Attack Point',
              y = 'Defense Point',
              data = pokemon,
              kind = 'kde')
```

```python
# Heatmap and correlation matrix


cormat = pokemon.iloc[::, 3:10].corr()
cormat
```

```python
plt.subplots(figsize = (8, 6))
sns.heatmap(cormat, vmax = 0.6, annot = True, fmt = '0.2f')
```

```python
# Linear Relations
```

```python
sns.lmplot(x = 'Speed Point',
           y = 'Defense Point',
           data = pokemon)
```

```python
sns.pairplot(pokemon,
             x_vars = ['Attack Point', 'Speed Point', 'Defense Point'],
             y_vars = ['Total Point'],
             kind = 'reg',
             height = 5)
```

# Categorical Data Analysis

```python
tips = sns.load_dataset('tips')
tips.head()

tips['size'].value_counts()
```

```python
# When ay data can be divided into different categories - Categorical columns

sns.stripplot(x = 'day',
              y = 'tip',
              data = tips,
              )
```

```python
sns.swarmplot(x='day',
              y='tip',
              data=tips)
```

```python
plt.subplots(figsize=(15,5))
sns.boxplot(x='day',
            y='tip'
          ,data=tips)
```

```python
# IQR - Inter Quartile Range - Q3-Q1
# 1st Quartile - 25% - Q1
# 2nd Quartile - 50% - Q2
# 3rd Quartile - 75% - Q3
# Mean, median, mode


Find the outliers - 1.5 X IQR - Q1
1.5 X IQR + Q3
```

```python
plt.subplots(figsize=(15,5))
sns.violinplot(x='day',
               y='tip',
               data=tips)
```

```python
plt.subplots(figsize = (15,5))
sns.barplot(x ='day',
            y ='tip',
            data = tips,)
```

```python
plt.subplots(figsize = (15,5))
sns.barplot(x ='day',
            y ='tip',
            data = tips,
            hue = 'size')
```
