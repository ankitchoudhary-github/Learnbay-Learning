```python
merge -> connecting the tables side by side
concatenating -> stacking the tables on top of each other
```

```python
f = open('/content/drive/MyDrive/datasets/customers.csv', 'w')
f.write('''CustID,Name
101,Riya
102,Arjun
103,Kabir''')
f.close()
```

```python
customers = pd.read_csv('/content/drive/MyDrive/datasets/customers.csv')
customers
```

```python
f = open('/content/drive/MyDrive/datasets/orders.csv', 'w')
f.write('''CustID,Amount
101,500
103,1200
101,300
104,900''')
f.close()
```

```python
orders = pd.read_csv('/content/drive/MyDrive/datasets/orders.csv')
orders
```

```python
result = pd.merge(customers, orders, on="CustID")
print(result)
```

```python
pd.merge(customers, orders, on="CustID", how="left")
```

```python
pd.merge(customers, orders, on="CustID", how="right")
```

```python
pd.merge(customers, orders, on="CustID", how="inner")
```

```python
pd.merge(customers, orders, on="CustID", how="outer")
```

  - inner = only what's in both (the overlap).
  - left = everything on the left, plus matches.
  - right = everything on the right, plus matches.
  - outer = everything from both sides.

# Datetime object

```python
f = open('/content/drive/MyDrive/datasets/orders.csv', 'w')
f.write('''
OrderID,OrderDate,Amount
1,2024-01-15,500
2,2024-02-20,800
3,2024-02-28,300
4,2024-03-10,950
''')
f.close()
```

```python
df = pd.read_csv('/content/drive/MyDrive/datasets/orders.csv', parse_dates=["OrderDate"])
df
```

```python
df.info()
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
print(df.dtypes)
```

```python
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month
df["Day"] = df["OrderDate"].dt.day
df["DayName"] = df["OrderDate"].dt.day_name()
print(df)
```
