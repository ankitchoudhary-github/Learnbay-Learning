```python
# Docstring
def calculate_discount(price, discount_pct):
    """
    Calculate the discounted price of a product.

    Args:
        price (float): Original price of the product in rupees.
        discount_pct (float): Discount percentage (0 to 100).

    Returns:
        float: Final price after applying the discount.

    Example:
        >>> calculate_discount(1000, 20)
        800.0
    """
    discount = price * (discount_pct / 100)
    return price - discount

# help(calculate_discount)
print(dir(calculate_discount))
print(calculate_discount.__doc__)
```

```python
# Multi Line comments
'''
This is a multi-line comment using single triple quotes.
It can span as many lines as you want.
Python ignores this entire block.
'''

"""
You can also use double triple quotes.
Both work the same way for multi-line comments.
Use whichever you prefer — just be consistent!
"""

# Multi-line comment for explaining a complex function
"""
Function: calculate_compound_interest
Purpose: Calculates the final amount after compound interest
Formula: A = P(1 + r/n)^(nt)
  P = principal, r = rate, n = compounds/year, t = years
"""
def calculate_compound_interest(P, r, n, t):
    return P * ((1 + r/n) ** (n*t))
```

```python
def check_grade(score):      # ← HEADER (ends with colon)
    if score >= 90:            # ← nested HEADER
        grade = "A"            # ← nested SUITE
    elif score >= 75:
        grade = "B"
    else:
        grade = "C"
    return grade
```

```python
def process_orders(orders):              # L0 header
    total = 0                             # L1 — function suite
    for order in orders:                  # L1 → L2 suite starts
        if order["status"] == "paid":    # L2 → L3 suite starts
            for item in order["items"]:  # L3 → L4 suite starts
                total += item["price"]   # L4 — innermost suite
    return total
```

```python
def generate_receipt(order):
    """Generate a Swiggy-style order confirmation receipt."""

    # \n — newline to separate sections
    # \t — tab to align columns
    # \\ — literal backslash (none needed here)
    # \" — literal quote inside double-quoted string

    header = (
        f"=== SWIGGY ORDER CONFIRMED ===\n"
        f"Order ID:\t#{order['id']}\n"         # \t aligns value
        f"Restaurant:\t{order['restaurant']}\n"
        f"Delivery to:\t{order['address']}\n"
        f"==============================\n"
    )

    items_section = "ITEMS ORDERED:\n"
    for item in order['items']:
        items_section += f"\t{item['qty']}x {item['name']}"   # \t indent
        items_section += f"\t\t₹{item['price']}\n"            # \t\t align price

    summary = (
        f"------------------------------\n"
        f"Subtotal:\t\t₹{order['subtotal']}\n"
        f"GST (5%):\t\t₹{order['gst']:.0f}\n"
        f"Delivery:\t\t₹{order['delivery']}\n"
        f"==============================\n"
        f"TOTAL:\t\t\t₹{order['total']:.0f}\n"
        f"------------------------------\n"
        f"\"Enjoy your meal!\" - Swiggy\n"   # \" for quotes in string
        f"ETA: {order['eta']} mins"
    )

    return header + items_section + summary

receipt = generate_receipt({
    "id":         "SWG-20240604-8821",
    "restaurant": "Saravana Bhavan",
    "address":    "HSR Layout, Bengaluru",
    "items": [
        {"qty": 2, "name": "Masala Dosa", "price": 120},
        {"qty": 1, "name": "Filter Coffee", "price": 60},
    ],
    "subtotal": 300, "gst": 15, "delivery": 0, "total": 315, "eta": 28
})
print(receipt)
```
