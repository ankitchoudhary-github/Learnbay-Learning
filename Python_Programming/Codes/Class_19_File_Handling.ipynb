```python
# Write mode - 'w'
# If the file exists, then it will first delete the content of the file after opening the file in write mode and places the cursor in the begining of the file.
# If the file does not exists then it will create a file with given bname and extension.
# In thios mode, You are allowed only to write into teh file but not allowed to read from the file.

f = open('demo.txt', 'w')
print('After opening the file, Current cursor Position is', f.tell())
print(f.write('Heelo Python File Handling class'))
print('After writing in the file, Current cursor Position is', f.tell())
f.close()

f = open('hello.txt', 'w')
print('After opening the file, Current cursor Position is', f.tell())
print(f.read())      # UnsupportedOperation: not readable
f.close()

# f = open('home.txt', 'w')
# f.close()
```

```python
# Append Mode - 'a'
# 1. If file exists then it opens the file in append mode and cursor will be at the end of the file
# If the file does not exists then it will create a file with given bname and extension.
# In this mode, you are allowed only to append but not allowed to read into the file

f = open('demo.txt', 'a')
print('After opening the file, Current cursor Position is', f.tell())
f.seek(0)
f.write('\n Hello\n')
print(f.tell())
f.close()
```

```python
# Read and write mode - 'r+'
# 1. If file exists then it opens the file in read mode and cursor will be at the begining of the file
# if the file does not exist then it raises FileNotFoundError
# In this mode, you are allowed read and write both.
# In read and write mode when we write the content then it replaces the content.


f = open('demo.txt', 'r+')
print('After opening the file, Current cursor Position is', f.tell())
print('-'*50)
print()
print('Content of the file before editing ------>')
print(f.read())
print()
print(f'before{f.tell()}')
f.seek(0)
print(f'after{f.tell()}')
f.write('Hello')
print('Content of the file after editing ------>')
f.seek(0)
print(f.read())
f.close()
```

```python
# Write and read mode - 'w+'
# If the file exists, then it will first delete the content of the file after opening the file in write mode and places the cursor in the begining of the file.
# If the file does not exists then it will create a file with given bname and extension.
# In this mode, you are allowed write and read both.
```

```python
f = open('demo.txt', 'w+')
print('After opening the file, Current cursor Position is', f.tell())
f.write('''# Write and read mode - 'w+'
# If the file exists, then it will first delete the content of the file after opening the file in write mode and places the cursor in the begining of the file.
# If the file does not exists then it will create a file with given bname and extension.
# In this mode, you are allowed write and read both.''')
print(f.tell())
print(f.read())
f.seek(0)
print(f.read())
f.close()
```

```python
# A hospital maintains a text file logging patient visits.
# Write a program that records a patient's visit by appending a line with the patient's id and
# their symptom (comma-separated) to the file patient_visits.txt,
# so each visit is saved on its own line for later reference.

def log_patient_visits(patient_id, pateint_name, symptoms):
  with open('patient_visits.txt', 'a') as file:
    file.write(f'{patient_id}, {pateint_name}, {symptoms}\n')

log_patient_visits(1, "Patient 1", "Diabetes")
log_patient_visits(2, "Patient 2", "Fever")
log_patient_visits(3, "Patient 3", "Asthma")
```
