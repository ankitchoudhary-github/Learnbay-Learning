```python
# Regex - Regular Expression

Regex is a powerful tool for identifying, matching, and replacing substrings within longer texts.
```

```python
text_to_search = '''bat AatA Cat c9T mat bat rat hat
abcdefghijklmnopqurtuvwxyz
ABCDEFGHIJKLMNOPQRSTUVWXYZ\s
321-555-4321
1234567890
Ha HaHa
MetaCharacters (Need to be escaped):
. ^ $ * + ? { } [ ] \ | ( )
guptaabhishek.com
abhishek-8765-gupta@yahoo.com
abhi_gyupta@gamil.abc
321-555-4321
123.555.1234
123*555*1234
123.555.1234
800-555-1234
900-555-1234
Mr. Schafer
Mr Smith
Ms Davis
Mrs. Robinson
Mr. T
Mr_hello
abhsihek.gupta@gmail.com
abhishek-1234gupta@gmail.org
abhishek@hotmail.net
at'''
```

```python
Regular expressions use metacharacters, special sequences, and quantifiers to match word characters, alphanumeric characters, or a single character.

# re module in python provides various methods to perfomr queries on an input string. They are -
# re.match()
# re.search()
# re.find()
# re.findall()
# re.split()
# re.sub()
# re.compile()
```

```python
# re.match(pattern, string) - It checks for match only at the begining of the string
# looks for the pattern in the string in the begining and retrusn a match object.
# To extract the value from match object, we need to use group() method on match object
# if patterns not found in the begining then it returns None.

pattern = r'\d[a-zA-Z]{2}'     # 9aB
res = re.match(pattern, text_to_search)
print(res)
print(res.group())
```

```python
# re.search(pattern, string) -
# looks for the pattern in the string and returns a match object of the first occurrenec of the pattern
# To extract the value from match object, we need to use group() method on match object
# if pattern s not found in the string then it returns None.

pattern = r'\d[a-zA-Z]{2}'     # 9aB
res = re.search(pattern, text_to_search)
print(res)
print(res.group())
```

```python
# findall(pattern, string) - It finds all the occurrences of the pattern and returns a
    # list of the all occurrences
# If the pattern is not found then it returns empty list

res = re.findall(r'a[^ab]{2}', text_to_search)
print(res)
```

```python
# finditer(pattern, string) -
# It finds all the occurrences of the pattern and returns an iterator of of match object of the all occurrences
# if the pattern is not found then returns none

res = re.finditer(r'a[^ab]{2}', text_to_search)
for val in res:
  print(val.group())

# ABCD12345E
```

```python
# re.sub(pattern, 'new_value', text_to_search) - Replaces parts of the string that match the pattern with another string.

# text = "My phone number is 123-456-7890"
updated_text = re.sub(r"^\d{3}-\d{3}-\d{4}$", "***-***-****", text_to_search)
print(updated_text)
```

```python
# re.split() -
text = "apple,banana;orange|grape"
fruits = re.split(r"[,;|]", text)
print(fruits)
```

```python
# Regular Expression pattern list
. -> Matches anything except newline
^ -> Matches the start of the string
$ -> Matches the end of the line
[...] -> Matches any single character in the bracket
[^...] -> Matches any single character not in the bracket
{n} -> Matches exactly n number of occurence of preceding expression
{n, m} -> Matches minimum n and maximum m number of occurence of preceding expression
x | y -> Matches either x or y
\d -> Matches digits[0-9]
\D -> Matches non-digits[^0-9]
\w -> Matches Alphanumeric characters - [a-zA-Z0-9_]
\W -> Matches Non-Alphanumeric characters - [^a-zA-Z0-9_]
\z - Matches end of the string
\G - Matches point where last match finished.
\b - Matches the empty string, but only at the begining and end of the word.
\B - Matches non-word boundary
\n, \t -> Matches new lines, carriage returns and tabs etc.
\s -> Matches white space
\S -> Matches non-white space
\A -> Matches the begining of the string
```

```python
# Syntax of regex
# Literals - Matching a specific characters - a, 1, !, %

# Metacharacters - Don't match themselves but they have a special meaning
  # - .(dot) - Matches any single charcter except a new line
  # - ^      - Matches the begining of the string
  # - $      - Matches the end of the string

    # Quantifiers - Specifiying the number of occurrences
  # - * - Matches zero or more occurrences
  # - + - Matches one or more occurrences
  # - ? - Matches zero or one occurrences
  # - {n} - Matches exactly n occurrences
  # - {n, } - Matches at least n occurrences
  # - {n, m} - Macthes between n and m occurrences


    # Special sequences of (Escape Characters)
  # - \d -> Matches any digit([0-9])
  # - \D -> Matches any non-digit([^0-9])
  # - \w - Matches any word character(Alphanumeric and underscore, [a-zA-Z0-9_])
  # - \W - Matches any non-word character([^a-zA-Z0-9_])
  # - \s - Matches any whitespace character
  # - \S - Matches any non-whitespace character
  # - \b - Matches a word boundary


    # Characters set - When we want to match one of a set of character
  #  - [ ]  - Matches any single character in the brackets (e.g [abc])
  #  - -    - Specifies a range(e.g [a-z], [0-9])
  #  - [^  ]- Matches any character not within the brackets

    # x | y -> Matches either x or y


  # Grouping and capturing
  # Paranthesis () -> It is used to group a part of the regex> This allows quantifiers to apply to the entire group
```

```python
# Find all the occurrences of word 'cat' where this word should be individual word in the string

content = 'cat catherine catholic wildcat cat copycat uncatachable cat'
pattern = r'\bcat\b'
res = re.finditer(pattern, content)
for ele in res:
  print(ele)
```

```python
text = "User IDs are user1, user2, and user3."
pattern = r"user\d" # Pattern for "user" followed by a digit

matches = re.findall(pattern, text)
print(matches)
```
