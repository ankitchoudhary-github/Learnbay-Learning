```python
Source Code - The code which we write in any PL - (Python - .py, C -> .c, C++ -> .cpp, Java -> .java etc)
Intermediate Code - The code which generated after translation(Compiler) - (Python - .pyc, c -> .exe, Java - .obj)
Machine Code - Binary code(0 and 1)
```

```python
# Compiler and Interpreter


Developer(10) - Source Code - Application - .py, .c, .java
Users of the application or software - Intermediate code - .pyc, .obj, .exe

Compiler - Compiles the whole source code at once and gives us the intermediate code after translation
Interpreter - Translates the source code line by line and does not gives us any intermediate code after translation
```

```python
# Scripting

# I want to add all the numbers which are odd below - get the final sum
lst = [10, 11, 12, 13, 14, 15, 16]
sum_res = 0
for ele in lst:
  if ele%2 != 0:
    sum_res += ele
print(sum_res)

lst = [21, 22, 23, 24, 25, 26]
sum_res = 0
for ele in lst:
  if ele%2 != 0:
    sum_res += ele
print(sum_res)
```

```python
# Procedural - User Defined Function

lst = [10, 11, 12, 13, 14, 15, 16]
def sum_res(my_lst):
  sum_res_odd = 0
  sum_res_even = 0
  for ele in my_lst:
    if ele%2 != 0:
      sum_res_odd += ele
    else:
      sum_res_even += ele
  return sum_res_odd
print(sum_res(lst))

lst = [21, 22, 23, 24, 25, 26]
print(sum_res(lst))
```
