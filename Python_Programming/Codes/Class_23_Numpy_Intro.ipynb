Arithmetic Operation on array


```python
# Arithmetic Operators on array perfomrs the operation elementwise
arr1 = np.random.randint(10, 20, 10)
print(arr1)


# Mathematical Operations

print(arr1+5)
print(arr1-5)
print(arr1*5)
print(arr1/5)
print(arr1//5)
print(arr1%5)
print(arr1**5)
```

```python
# Comparison operations
print(arr1>5)
print(arr1<5)
print(arr1<=5)
print(arr1>=5)
print(arr1==5)
print(arr1!=5)
print(arr1&5)
```

```python
# To perform these operation on the array, the dimension and shape of the array should remain same

arr1 = np.random.randint(10, 90, 5)
arr2 = np.random.randint(10, 90, 5)
print(arr1)
print(arr2)
```

```python
# Arithmetic operations - element wise
print(arr2+arr1)
print(arr2-arr1)
print(arr2*arr1)
print(arr2/arr1)
print(arr2//arr1)
print(arr2%arr1)
print(arr2**arr1)
```

```python
print(arr1 > arr2)
print(arr1 < arr2)
print(arr1 >= arr2)
print(arr1 <= arr2)
print(arr1 == arr2)
print(arr1 != arr2)
print(arr1 & arr2)
```

```python
arr1[arr1 > arr2]
```

```python
# Vector product(dot product) - np.dot(arr1, arr2)
# No of elements in both the array shgould be same
print(arr1)
print()
print(arr2)
```

```python
np.dot(arr1, arr2)
```
