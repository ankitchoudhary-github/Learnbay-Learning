```python
# Data Types in Python
# Fundamental Data Types - Integers, Float, Complex, Boolean, String, None

# Integer - Any whole number without decimal point
# May Representation - Decimal, Binary, Octal, Hexadecimal
# Only integers have these 4 type of representation

# Decimal(10) - Allowed digits are - 0-9, - Default representation of number we use in our day to day life - 23, 45, 90, -98, 9879871263865127, No prefix required
# Binary(2) Representation - Allowed digits are - 0 and 1, bin(num) -> output will be in binary format, Prefix - 0b, 0B, - 0b10101, 0b110011
# Octal(8) Representation - Allowed digits are - 0 to 7, oct(num) -> output will be in octal format, Prefix - 0o, 0O, - 0o5674, 0O234156
# Hexadecimal(16) Representation - Allowed digits are - 0 to 9 and a/A to f/F, hex(num) -> output will be in hexadecimal format, Prefix - 0x, 0X, - 0xface, 0x45acd

dec_num = 987654
print('Decimal to Binary - ', bin(dec_num))
print('Decimal to Octal - ', oct(dec_num))
print('Decimal to Hexadecimal - ', hex(dec_num))
print()
print('Binary to Decimal - ', 0b11110001001000000110)
print('Binary to Octal - ', oct(0b11110001001000000110))
print('Binary to Hexadecimal - ', hex(0b11110001001000000110))
print()
print('Octal to Decimal - ', 0o3611006)
print('Octal to Binary - ', bin(0o3611006))
print('Octal to Hexadecimal - ', hex(0o3611006))
print()
print('Hexadecimal to Decimal - ', 0xf1206)
print('Hexadecimal to Binary - ',bin(0xf1206))
print('Hexadecimal to Octal - ', oct(0xf1206))
```

```python
# Mutable(Modify the data and changes will take place in same object) and Immutable(We can not modify the data and if we will try to do so then it will create a new object)

# Mutable - List(Indexing), set, dictionary - Item Assignment is Possible
# Immutable - String, Tuple(Indexing), Frozenset - Item Assignment is not possible

# String
my_str = 'This is Python class'

# For all the sequence type -

# They supports Positive and negative indexing both
# +ve index starts from left side and -ve index start from right side
# They support basic operation like Indexing, Slicing, Concatenation, Repetition and Membership

# String is Immutable
# my_str = 'This is Python class'
# my_str[0] = 't'

# List is Mutable = Item Assignment is possible
my_lst = [1, 2, 3, 4]
print('Before Operation')
print(my_lst, id(my_lst))
my_lst[-1] = 40
print('After Operation')
print(my_lst, id(my_lst))

```
