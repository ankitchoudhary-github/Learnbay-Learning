```python
# Generating a list of numbers using normal function concept


def get_numbers_list(num):
  out_numbers = []
  for ele in range(num):
    out_numbers.append(ele)
  return out_numbers

# get_numbers_list(10)

# get_numbers_list(10000000000000)
```

```python
# Generating a list of numbers using generator concept

def get_numbers_generators(num):
  for ele in range(num):
    yield ele

# print(dir(get_numbers_generators))

my_gen = get_numbers_generators(1000000000000000000000000000)
# print(dir(my_gen))

print(next(my_gen))
print(next(my_gen))
print(next(my_gen))
print(next(my_gen))
```

```python
# Tuple comprehension - similar to list comprehgension

# Tuple comprehension gives us a generator object

# (expr for loop)
# (expr for loop if cond)
# (expr if cond else expr for loop)

tpl_gen = (i for i in range(10))
print(tpl_gen, type(tpl_gen))
print(next(tpl_gen))
```

```python
def unique_id_gen():
  user_num = 1
  while True:
    yield f"user_{user_num}"
    yield user_num
    user_num += 1

unique_id = unique_id_gen()
print(next(unique_id))
print(next(unique_id))
print(next(unique_id))
```

```python
# Factorial of a number -> 5! -> 5 X 4 X 3 X 2 X 1 = 120 -> num = 5 => 5X4X3 => num * (num-1)'
# Base case = 1
# 0! = 1
# 5! = 5 X 4! = 5 X 4 X 3! = 5 X 4 X 3 X 2! = 5 X 4 X 3 X 2 X 1!

def my_fact(num):
  if num < 0:
    return 'Enter a valid number'
  elif num == 0:
    return 1
  elif num == 1:
    return 1
  else:
    return num * my_fact(num-1)
my_fact(5)
```

```python
# [1, [2, 3], [4, [5]]] -> find the sum of all elements in the nested list

def total_sum(data_lst):
  total = 0
  for ele in data_lst:
    if isinstance(ele, list):
      total += total_sum(ele)
    else:
      total += ele
  return total
total_sum([1, [2, 3], [4, [5]]])
```

```python
# clean_data.py

"""
UrbanCart Ask-AI Assistant — Stage 4: Data Cleaning Pipeline

Cleans the messy product catalog from Stage 2A into standardized fields,
logging every transformation to an audit trail. Cleaned columns are added
ALONGSIDE the original raw columns (never overwritten) — so the raw source
value is always recoverable if a cleaning rule turns out to be wrong.

Design principle (per Stage 2/3's findings): never silently drop a row or
leave a field blank. Every field always gets SOME value — a cleaned value,
a sensible default (e.g. "Other", "Unknown"), or an explicit None that gets
logged — and records with many corrections are flagged for human review
rather than trusted blindly.
"""

import logging
import re

import pandas as pd
from rapidfuzz import fuzz, process

logging.basicConfig(
    filename="logs/catalog_cleaning_audit.log",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
)

CATEGORY_MAPPING = {
    "Fashion": "Fashion", "fashion": "Fashion", "Apparel": "Fashion", "Clothing": "Fashion",
    "Men Fashion": "Fashion", "Women Fashion": "Fashion", "Ethnic Wear": "Fashion",
    "Electronics": "Electronics", "electronics": "Electronics", "Mobiles & Accessories": "Electronics",
    "Gadgets": "Electronics", "Audio": "Electronics",
    "Home & Kitchen": "Home & Kitchen", "home": "Home & Kitchen", "Home Decor": "Home & Kitchen",
    "Kitchen Appliances": "Home & Kitchen",
    "Footwear": "Footwear", "Shoes": "Footwear", "Sandals": "Footwear",
}

# Text ratings are mapped to an APPROXIMATE numeric band only for internal
# filtering/flagging — never presented to the LLM as a precise real rating
# (that distinction is preserved via each note's "APPROXIMATE" tag).
RATING_TEXT_BANDS = {
    "excellent": 4.5,
    "good": 4.0,
    "poor": 2.0,
}


def clean_price(price_value):
    """
    Standardize a messy price string into a clean integer (INR).
    Handles all 5 formats from the raw catalog: '1999', '\u20b91999',
    '\u20b91,999', '1.9k', 'Rs 1999/-'.

    Returns (cleaned_value: int | None, was_transformed: bool, note: str)
    """
    if pd.isna(price_value):
        return None, False, "missing_value"

    original = price_value
    price_str = str(price_value).strip()

    if re.match(r"^\d+$", price_str):
        return int(price_str), False, "already_clean"

    if price_str.endswith("k"):
        try:
            cleaned = round(float(price_str[:-1]) * 1000)
            return cleaned, True, f"parsed_k_suffix_format ({original} -> {cleaned})"
        except ValueError:
            return None, True, f"unparseable_k_suffix_format ({original})"

    digits_only = re.sub(r"[^\d]", "", price_str)
    if not digits_only:
        return None, True, f"unparseable_price_format ({original})"

    cleaned = int(digits_only)
    return cleaned, True, f"stripped_symbols ({original} -> {cleaned})"


def standardize_category(category_value):
    """
    Map a messy category string to one of:
    ['Fashion', 'Electronics', 'Home & Kitchen', 'Footwear', 'Other'].

    Returns (standardized_value: str, was_transformed: bool, note: str)
    """
    if pd.isna(category_value):
        return "Other", True, "missing_value_defaulted_to_other"

    category_str = str(category_value).strip()

    if category_str in CATEGORY_MAPPING:
        standardized = CATEGORY_MAPPING[category_str]
        was_transformed = standardized != category_str
        note = "already_clean" if not was_transformed else f"mapped_known_variant ({category_str} -> {standardized})"
        return standardized, was_transformed, note

    # Fuzzy match fallback for variants not explicitly in our mapping — e.g.
    # a seller-entered typo like "Electroncis" or "Foot Wear"
    match = process.extractOne(category_str, CATEGORY_MAPPING.keys(), scorer=fuzz.ratio)
    if match and match[1] >= 80:
        standardized = CATEGORY_MAPPING[match[0]]
        return standardized, True, f"fuzzy_matched ({category_str} -> {standardized}, score={match[1]:.0f})"

    return "Other", True, f"unrecognized_category_defaulted_to_other ({category_str})"


def standardize_stock_status(stock_value, stock_quantity=None):
    """
    Reconcile the free-text stock_status against numeric stock_quantity into
    one of: ['In Stock', 'Low Stock', 'Out of Stock', 'Unknown'].

    Quantity is treated as the more reliable signal when present — sellers
    and legacy systems enter the free-text status inconsistently, but a
    quantity count is a harder number to get wrong.

    Returns (standardized_value: str, was_transformed: bool, note: str)
    """
    if stock_quantity is not None and not pd.isna(stock_quantity):
        if stock_quantity < 0:
            return "Unknown", True, f"invalid_negative_quantity ({stock_quantity})"
        if stock_quantity == 0:
            standardized = "Out of Stock"
        elif stock_quantity <= 5:
            standardized = "Low Stock"
        else:
            standardized = "In Stock"

        was_transformed = str(stock_value).strip() != standardized
        note = (
            "already_clean" if not was_transformed
            else f"reconciled_from_quantity ({stock_value} + qty={stock_quantity} -> {standardized})"
        )
        return standardized, was_transformed, note

    # No usable quantity — fall back to interpreting the free-text label
    if pd.isna(stock_value):
        return "Unknown", True, "missing_status_and_quantity"

    status_str = str(stock_value).strip().lower()

    if status_str in ("in stock", "available"):
        return "In Stock", True, f"normalized_text_label ({stock_value})"
    if "left" in status_str:
        return "Low Stock", True, f"normalized_text_label ({stock_value})"
    if status_str in ("out of stock", "oos", "sold out", "coming soon"):
        return "Out of Stock", True, f"normalized_text_label ({stock_value})"

    return "Unknown", True, f"unrecognized_status_label ({stock_value})"


def clean_rating(rating_value):
    """
    Return a valid float rating (1.0-5.0), or None if it can't be determined.

    Returns (cleaned_value: float | None, was_transformed: bool, note: str)
    """
    if pd.isna(rating_value):
        return None, False, "missing_value"

    rating_str = str(rating_value).strip()

    try:
        rating_num = float(rating_str)
        if 1.0 <= rating_num <= 5.0:
            return rating_num, False, "already_clean"
        return None, True, f"out_of_range_discarded ({rating_str})"
    except ValueError:
        pass

    lowered = rating_str.lower()

    stars_match = re.match(r"^(\d+(\.\d+)?)\s*stars?$", lowered)
    if stars_match:
        extracted = float(stars_match.group(1))
        if 1.0 <= extracted <= 5.0:
            return extracted, True, f"extracted_from_stars_text ({rating_str} -> {extracted})"

    for keyword, estimated_value in RATING_TEXT_BANDS.items():
        if keyword in lowered:
            return estimated_value, True, f"estimated_from_text ({rating_str} -> ~{estimated_value}, APPROXIMATE)"

    return None, True, f"unparseable_rating_discarded ({rating_str})"


def run_cleaning_pipeline(df):
    """
    Apply all cleaning functions and return:
    - df_clean: original dataframe plus new *_clean columns
    - df_audit: one row per field-level correction made (a record with 3
      corrected fields produces 3 audit rows, not 1)

    Records with more than 3 corrected fields are flagged via
    df_clean['needs_manual_review'] for the Catalog Ops team.
    """
    df_clean = df.copy()
    audit_rows = []

    cleaned_prices, cleaned_categories, cleaned_stock, cleaned_ratings = [], [], [], []
    fields_changed_per_row = []

    for _, row in df.iterrows():
        changed_fields = 0

        price, price_changed, price_note = clean_price(row["price"])
        cleaned_prices.append(price)
        if price_changed:
            changed_fields += 1
            audit_rows.append({"sku": row["sku"], "field": "price", "original_value": row["price"], "cleaned_value": price, "note": price_note})
            logging.info(f"sku={row['sku']} | field=price | {price_note}")

        category, category_changed, category_note = standardize_category(row["category"])
        cleaned_categories.append(category)
        if category_changed:
            changed_fields += 1
            audit_rows.append({"sku": row["sku"], "field": "category", "original_value": row["category"], "cleaned_value": category, "note": category_note})
            logging.info(f"sku={row['sku']} | field=category | {category_note}")

        stock, stock_changed, stock_note = standardize_stock_status(row["stock_status"], row["stock_quantity"])
        cleaned_stock.append(stock)
        if stock_changed:
            changed_fields += 1
            audit_rows.append({"sku": row["sku"], "field": "stock_status", "original_value": row["stock_status"], "cleaned_value": stock, "note": stock_note})
            logging.info(f"sku={row['sku']} | field=stock_status | {stock_note}")

        rating, rating_changed, rating_note = clean_rating(row["rating"])
        cleaned_ratings.append(rating)
        if rating_changed:
            changed_fields += 1
            audit_rows.append({"sku": row["sku"], "field": "rating", "original_value": row["rating"], "cleaned_value": rating, "note": rating_note})
            logging.info(f"sku={row['sku']} | field=rating | {rating_note}")

        fields_changed_per_row.append(changed_fields)

    df_clean["price_clean"] = cleaned_prices
    df_clean["category_clean"] = cleaned_categories
    df_clean["stock_status_clean"] = cleaned_stock
    df_clean["rating_clean"] = cleaned_ratings
    df_clean["fields_corrected_count"] = fields_changed_per_row
    df_clean["needs_manual_review"] = df_clean["fields_corrected_count"] > 3

    df_audit = pd.DataFrame(audit_rows)

    return df_clean, df_audit


if __name__ == "__main__":
    df_raw = pd.read_csv("data/raw/urbancart_product_catalog_raw.csv")
    df_clean, df_audit = run_cleaning_pipeline(df_raw)

    df_clean.to_csv("data/processed/urbancart_product_catalog_clean.csv", index=False)
    df_audit.to_csv("logs/catalog_cleaning_audit.csv", index=False)

    print(f"Cleaned {len(df_clean)} records")
    print(f"Audit trail: {len(df_audit)} field-level corrections logged")
    print(f"Records flagged for manual review (>3 fields corrected): {df_clean['needs_manual_review'].sum()}")

    # Recompute Data Quality Score on the Tier-1 fields using the CLEANED
    # columns, to show the improvement over Stage 3B's 52.7% baseline
    required_clean = ["category_clean", "stock_status_clean", "price_clean"]
    dq_score_clean = (~df_clean[required_clean].isnull().any(axis=1)).mean() * 100
    print(f"Data Quality Score (Tier-1 fields, post-cleaning): {dq_score_clean:.1f}%")
```
