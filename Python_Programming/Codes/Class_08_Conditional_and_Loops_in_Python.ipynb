```python
# Write a program that prompts the user to input number of calls and calculate the
# monthly telephone bills as per the following rule:
# Minimum Rs. 200 for up to 100 calls.
# Plus Rs. 0.60 per call for next 50 calls.
# Plus Rs. 0.50 per call for next 50 calls.
# Plus Rs. 0.40 per call for any call beyond 200 calls.

no_of_calls = int(input('Enter the number of calls: '))
# total_bill = 0.0
if (no_of_calls >= 0) and (no_of_calls <= 100):
  total_bill = 200
  print(f'Total bill amount for {no_of_calls} calls is {total_bill}')
elif (no_of_calls >= 101) and (no_of_calls <= 150):
  total_bill = 200 + (no_of_calls - 100) * 0.6
  print(f'Total bill amount for {no_of_calls} calls is {total_bill}')
elif (no_of_calls >= 151) and (no_of_calls <= 200):
  total_bill = 200 + (50 * 0.60) + (no_of_calls - 150) * 0.50
  print(f'Total bill amount for {no_of_calls} calls is {total_bill}')
elif (no_of_calls >= 201):
  total_bill = 200 + (50 * 0.60) + (50 * 0.50) + (no_of_calls - 200)*0.40
  print(f'Total bill amount for {no_of_calls} calls is {total_bill}')
else:
  print('Input is wrong')
```

# User Permission and Access Control

Determine if a user can access a specific feature based on their **User Role**, **Subscription Tier**, **Account Status**, and **Feature Sensitivity**.

Question

1. An application controls feature access. A user must satisfy multiple checks:

2. Role: Only 'Admin' or 'Editor' can proceed. 'Viewer' is denied.

3. Subscription: For 'Admin'/'Editor', only 'Premium' or 'Enterprise' can proceed. 'Basic' is denied.

4. Status: The account must be 'Active'. 'Suspended' is denied.

5. Sensitivity: If the feature is 'Critical', only 'Admin' is allowed. 'Editor' is denied for 'Critical' features. 'Standard' features are open to both.



```python
role = input('Enter Role [Admin, Editor, Viewer]: ').capitalize()
subscription = input('Enter Subscription type [Premium, Enterprise]: ').capitalize()
status = input('Enter Status [Active, Suspended]: ').capitalize()
feature = input('Enter Feature [Standard, Critical]: ').capitalize()
if role in ['Admin', 'Editor']:
  if subscription in ['Premium', 'Enterprise']:
    if status == 'Active':
      if feature == 'Standard':
        print('Access Granted')
      elif feature == 'Critical':
        if role == 'Admin':
          print('Access Granted')
        else:
          print('Access is denied')
      else:
        print('Access is denied')
    else:
      print('Access is denied')
  else:
    print('Access is denied')
else:
  print('Access is denied')
```

```python
# range() - used to create a range object
# range object - It is used to create a sequence of integers
# range function returns the range object - (generator object)

# Syntax 1
# range(start, stop ,step) where start, stop and step has same meaning as string slicing concepts
# Stop is exclusive - whatever stop value we will give it wull generate the numbers till stop-1
# Step is optional, default value of step is 1
# Step can be +ve and -ve both'
# When step is +ve - then numbers will be generated from left to right in given range as per start and stop
# When step is -ve - then numbers will be generated from right to left in given range as per start and stop
# Default start is 0

# Syntax 2
# range(stop) - # range(stop) -> It will generate the sequence of numbers from 0 to (stop-1)

# 9 to -1 -> step -ve
# -9 to 0 -> step is +ve

# range function always returns range object so you can not see the values directly, so we use the concept of realization of the values

# print(list(range(5)))
# print(tuple(range(5)))
# print(list(range(1, 11)))
# print(list(range(1, 11, 2)))
# print(tuple(range(2, 11, 2)))
# print(tuple(range(1, 11, -1)))
# print(tuple(range(10, -10, -2)))
# print(tuple(range(10, -10, 2)))
```
