```python

from openai import OpenAI

MODEL = "qwen2.5:3b"

client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

question = "Suggest a hill station near Bengaluru."

# System instruction comes at the top
messages = [
    {
        "role": "system",
        "content": "You are a helpful travel assistant."
    },
    {
        "role": "user",
        "content": question
    }
]

response = client.chat.completions.create(
    model=MODEL,
    messages=messages
)

final_resp = response.choices[0].message.content

print("Generated text:", final_resp)
```

Role - who The AI should act as? "You are a helpful and polite banking assistant"

Task - The exact job - Action Very - Summarise, Translate, Classify

Context
  1. Situational Context - "This reply is for a customer who has complained twice already and is very upset."
  2. Data Context - From where the model will answer the query.

 Constraints - The rules and the limits that we set for the model - ANswer in 50 words, or explain in 2-3 line, Do not give any medical advice, Do not make up any fact or suggest any made up interest on looan

 Output format - Bullet points, Text format, JSON or XML format

```python

import os
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()

API_KEY = os.getenv("GEMINI_API_KEY")
MODEL = "gemini-3.1-flash-lite"

client = genai.Client(api_key=API_KEY)

messages = [
    types.Content(
        role="user",
        parts=[
            types.Part.from_text(
                text="Suggest a hill station near Bengaluru."
            )
        ]
    )
]

response = client.models.generate_content(
    model=MODEL,
    contents=messages,
    config=types.GenerateContentConfig(
        system_instruction="You are a helpful travel assistant.",
        max_output_tokens=100,
        temperature=1.0
    )
)

print("Generated text:", response.text)
```
