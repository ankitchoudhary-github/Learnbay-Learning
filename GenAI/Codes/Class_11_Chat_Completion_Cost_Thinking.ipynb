```python

import os
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()

API_KEY = os.getenv("GEMINI_API_KEY")
MODEL = "gemini-3.1-flash-lite"

client = genai.Client(api_key=API_KEY)

messages = [
    types.Content(
        role="user",
        parts=[
            types.Part.from_text(
                text="Suggest a hill station near Bengaluru."
            )
        ]
    )
]

response = client.models.generate_content(
    model=MODEL,
    contents=messages,
    config=types.GenerateContentConfig(
        system_instruction="You are a helpful travel assistant.",
        max_output_tokens=100,
        temperature=1.0
    )
)

print("Generated text:", response.text)
```

## Explicit Caching Configuration
For standard implicit caching, OpenAI handles everything automatically once your prompt prefix exceeds 1,024 tokens. However, to manage explicit control or disable automatic caching behavior, you configure your request body or SDK call as follows: [2, 3, 4, 5]

* Set prompt_cache_options to mode: "explicit" in your model parameters. [2, 3]
* Add a prompt_cache_breakpoint directly inside your stable content blocks or messages (such as large system instructions or static prefixes) to explicitly declare where the cache snapshot should occur. [2, 3]
* Use an optional prompt_cache_key if you need to namespace or force alignment for specific recurring application states or sessions. [1, 3]

```python

from openai import OpenAI
import time

# Connect to the local Ollama API using the OpenAI SDK
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

# SwiftBank rules: the stable system prompt
BANK_RULES = """
You are SwiftBank's assistant. Answer briefly and professionally.

DATA PRIVACY:
- Never request passwords, PINs, or OTPs.
- Do not reveal confidential customer information.
- Protect personal and financial information.

SECURITY:
- Never claim to execute a banking transaction.
- Never claim to block a card or access an account.
- Direct customers to official SwiftBank channels for account actions.
- Escalate complex or sensitive requests to a human agent.

ACCOUNTABILITY:
- Do not invent fees, interest rates, or bank policies.
- If information is missing, direct customers to SwiftBank's
  official fee schedule or customer support.
- Provide general informational guidance only.

RESPONSE STYLE:
- Answer clearly and briefly.
- Never request sensitive banking credentials.
"""

# Customer questions
q1 = "What is the fee for a late credit card payment?"
q2 = "How do I block a lost debit card?"


def ask_bank(question):
    start_time = time.perf_counter()

    response = client.chat.completions.create(
        model="qwen2.5:3b",
        messages=[
            {"role": "system", "content": BANK_RULES},
            {"role": "user", "content": question}
        ],
        max_tokens=150
    )

    elapsed_time = time.perf_counter() - start_time

    # Display the generated answer
    print("\n" + "=" * 50)
    print("QUESTION:", question)
    print("ANSWER:", response.choices[0].message.content)

    # Display token usage
    usage = response.usage

    print("\nInput tokens:", usage.prompt_tokens)
    print("Output tokens:", usage.completion_tokens)
    print("Total tokens:", usage.total_tokens)

    # Display cached prompt tokens, if reported by Ollama
    details = getattr(usage, "prompt_tokens_details", None)
    cached_tokens = getattr(details, "cached_tokens", None)

    if cached_tokens is not None:
        print("Cached prompt tokens:", cached_tokens)
    else:
        print("Cached prompt tokens: Not reported by this Ollama version")

    print("Response time:", round(elapsed_time, 2), "seconds")


# Run both requests with the same stable system prompt
ask_bank(q1)
ask_bank(q2)
```
