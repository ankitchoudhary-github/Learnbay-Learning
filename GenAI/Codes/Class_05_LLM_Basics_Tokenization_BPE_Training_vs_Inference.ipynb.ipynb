```python
from google import genai
from dotenv import load_dotenv
import os

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)
```

```python
text = "Hello, world!"

response = client.models.count_tokens(
    model="gemini-3.8-flash",
    contents=text
)

print("Total tokens:", response.total_tokens)
```
