```python
import os
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()

API_KEY = os.getenv("GEMINI_API_KEY")
MODEL = "gemini-3.1-flash-lite"

client = genai.Client(api_key=API_KEY)

response = client.models.generate_content(
    model=MODEL,
    contents="Complete this sentence with exactly one word: I love eating",
    config=types.GenerateContentConfig(
        max_output_tokens=10,
        temperature=1.0
    )
)

print("Generated text:", response.text)
```

def show_probability_menu(top_k=8):

    response = client.chat.completions.create(
        model=MODEL,
        max_tokens=1,
        temperature=1.0,
        logprobs=True,
        top_logprobs=min(top_k, 5),
        messages=[{"role": "user", "content": PROMPT}],
    )

    final_resp = response.choices[0].message.content

    top_candidates = response.choices[0].logprobs.content[0].top_logprobs

    for rank, candidate in enumerate(top_candidates, start=1):
        probability_pct = math.exp(candidate.logprob) * 100

        print(
            f"Rank {rank} -> Tokens {candidate.token} "
            f"and Probability Percent -> {probability_pct:.4f}%"
        )

    print(f'Chosen Token is {final_resp}')

1. AS we know that LLM is next token predictor.
2. LLM generates a list of Probable tokens with probability score
3. We can not consider all the probable tokens to be selcted by the LLM randomly
4. We do the sampling - top_k sampling - where we provide a number like 5 or 8 or 10 and then model will select top 5 oir 8 or 10 tokens as next tokens and will be predicting the next tokens from there
5. top_p sampling - percenteage 99% -> it will combine the probability score for all the tokens till it reaches the specified probability under top_p parameter and then selection will be happening from there
6. temerateure - out of selected sample which token now has to be selected - 0 to 1 for majority models whereas in OPenAI 0 to 2
7. Setting the value near to 0 will always result in same selction of highest probability token and setting to near 1 will be generationg creative respinse

# API - Application Programming Interface

# Used in app communication

# Client Server model or Request Respinse model

# Parts of an API call

1. The address (endpoint):-
2. The method: - GET(Fetching some information) and POST(send something to be processed)
3. The API key - Password
4. The body: The actual content of your request


```python
import requests

url = "https://api.openai.com/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

body = {
    "model": "gpt-4o-mini",
    "messages": [
        {"role": "user", "content": "Write a one-line description for a blue cotton kurta."}
    ]
}

response = requests.post(url, headers=headers, json=body)

print(response.json())
```
