```python
# A library called "tiktoken" counts tokens the way GPT models do.
import tiktoken

# Load the tokenizer (the "chopping tool") for a model.
encoder = tiktoken.get_encoding("cl100k_base")

text = "the cat sat on the mat"

# .encode() turns text into a list of token ID numbers.
tokens = encoder.encode(text)

print(tokens)
print(len(tokens))
```
