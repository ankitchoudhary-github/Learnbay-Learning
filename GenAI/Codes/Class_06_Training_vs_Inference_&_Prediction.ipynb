```python
from google import genai
from google.genai import types
from dotenv import load_dotenv
import os

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="In how many days I can return my shoes?",
    config=types.GenerateContentConfig(
        automatic_function_calling=types.AutomaticFunctionCallingConfig(
            disable=True
        )
    )
)

print(response.text)
```

# LLM - Next Toekn Predictor -
# An LLM does one job, over and over: given text so far, guess the single most likely next word.

from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

result = generator("I take my tea with", max_new_tokens=1)

print(result[0]["generated_text"])

Look at the text so far.
Predict the single most likely next word (token).
Add that word to the text.
Look at the new, longer text — and predict the next word again.
Repeat, over and over, until it decides to stop (or hits a limit).
